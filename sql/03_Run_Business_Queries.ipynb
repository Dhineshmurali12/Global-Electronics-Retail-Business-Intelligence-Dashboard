# 2. Customer Analysis

## Objective

The purpose of this section is to analyze the customer base across different regions and demographic groups.

The analysis answers the following business questions:

- Which countries have the highest number of customers?
- What is the gender distribution of customers?
- Which continents contribute the most customers?

These insights help businesses identify major customer markets and support targeted marketing and expansion strategies.

```python
query = """
SELECT
    Country,
    COUNT(*) AS Total_Customers
FROM customers
GROUP BY Country
ORDER BY Total_Customers DESC;
"""

pd.read_sql(query, connection)
```

### Business Insight

This analysis identifies the countries with the largest customer base.
Countries with a higher number of customers represent important markets for business growth, marketing campaigns, and sales expansion.

```python
query = """
SELECT
    Gender,
    COUNT(*) AS Total_Customers
FROM customers
GROUP BY Gender;
"""

pd.read_sql(query, connection)
```

### Business Insight

Understanding gender distribution provides demographic insights that can help the company design personalized marketing campaigns and improve customer engagement.

```python
query = """
SELECT
    Continent,
    COUNT(*) AS Total_Customers
FROM customers
GROUP BY Continent
ORDER BY Total_Customers DESC;
"""

pd.read_sql(query, connection)
```

### Business Insight

This analysis highlights the geographic distribution of customers across continents.
It helps identify the strongest regional markets and supports international business expansion decisions.

# 3. Product Analysis

## Objective

This section analyzes the company's product portfolio by examining product categories, brands, pricing, and other product attributes.

The analysis answers questions such as:

- Which product categories contain the most products?
- Which brands offer the largest product portfolio?
- What is the average selling price of products?
- What is the average product cost?
- Which product colors are most common?

These insights help understand the company's product strategy, pricing structure, and inventory composition.

## Business Question 1

### Which product categories contain the highest number of products?

```python
query = """
SELECT
    Category,
    COUNT(*) AS Total_Products
FROM products
GROUP BY Category
ORDER BY Total_Products DESC;
"""

pd.read_sql(query, connection)
```

### Business Insight

This analysis identifies which product categories make up the largest portion of the company's product portfolio. Categories with more products indicate a stronger business focus and broader customer offerings.

## Business Question 2

### Which brands offer the largest number of products?

```python
query = """
SELECT
    Brand,
    COUNT(*) AS Total_Products
FROM products
GROUP BY Brand
ORDER BY Total_Products DESC;
"""

pd.read_sql(query, connection)
```

### Business Insight

Brands with a larger product portfolio contribute significantly to the company's offerings and may require focused inventory management and marketing efforts.## Business Question 3

### What is the average selling price for each product category?

```python
query = """
SELECT
    Category,
    ROUND(
        AVG(
            CAST(REPLACE(REPLACE("Unit Price USD",'$',''),',','') AS REAL)
        ),
        2
    ) AS Average_Selling_Price
FROM products
GROUP BY Category
ORDER BY Average_Selling_Price DESC;
"""

pd.read_sql(query, connection)
```

### Business Insight

Average selling price helps identify premium and budget product categories. Premium categories typically generate higher revenue per sale, while lower-priced categories may rely on higher sales volume.

## Business Question 4

### What is the average product cost?

```python
query = """
SELECT
    ROUND(
        AVG(
            CAST(REPLACE(REPLACE("Unit Cost USD",'$',''),',','') AS REAL)
        ),
        2
    ) AS Average_Product_Cost
FROM products;
"""

pd.read_sql(query, connection)
```

### Business Insight

The average product cost provides an overview of the company's investment in inventory and serves as a baseline for future profitability analysis.

## Business Question 5

### Which product colors are most common?

```python
query = """
SELECT
    Color,
    COUNT(*) AS Total_Products
FROM products
GROUP BY Color
ORDER BY Total_Products DESC;
"""

pd.read_sql(query, connection)
```

### Business Insight

Analyzing product color distribution helps understand product design preferences and inventory composition. Popular colors may reflect customer demand and market trends.

# 4. Store Analysis

- Which countries have the highest number of stores?

- What is the average store size?

## Business Question 1

### Business Insight

This analysis identifies countries with the largest retail presence. Markets with more stores typically represent the company's strongest operational regions.

## Business Question 2

### Which states have the highest number of stores?

### Business Insight

The distribution of stores across states helps identify regional concentration and supports decisions regarding expansion and resource allocation.

## Business Question 3

### What is the average store size?

```python
pd.read_sql("PRAGMA table_info(stores);", connection)
```

```python
query = """
SELECT
ROUND(AVG("Square Meters"),2) AS Average_Store_Size
FROM stores;
"""

pd.read_sql(query, connection)
```

## Business Question 4

### In which years were the stores opened?

```python
query = """
SELECT
SUBSTR("Open Date",7,4) AS Opening_Year,
COUNT(*) AS Stores_Opened
FROM stores
GROUP BY Opening_Year
ORDER BY Opening_Year;
"""

pd.read_sql(query, connection)
```

### Business Insight

This analysis shows the company's expansion over time by identifying the years in which new stores were opened. It helps understand historical growth patterns and investment periods.

# 5. Sales Analysis

## Objective

This section combines the Sales and Products tables using SQL JOINs to analyze sales performance.

The analysis answers the following business questions:

- Which products sell the highest quantity?
- Which brands generate the highest sales volume?
- Which product categories contribute the most sales?

This section demonstrates relational database analysis using INNER JOIN, a fundamental SQL skill for Data Analysts.

```python
query = """
SELECT
    p."Product Name",
    SUM(s.Quantity) AS Total_Quantity_Sold
FROM sales s
INNER JOIN products p
ON s.ProductKey = p.ProductKey
GROUP BY p."Product Name"
ORDER BY Total_Quantity_Sold DESC
LIMIT 10;
"""

pd.read_sql(query, connection)
```

### Business Insight

The top-selling products contribute significantly to overall sales volume. Identifying these products helps improve inventory planning, forecasting, and promotional strategies.

## Business Question 2

### Which brands generate the highest sales volume?

```python
query = """
SELECT
    p.Brand,
    SUM(s.Quantity) AS Total_Quantity_Sold
FROM sales s
INNER JOIN products p
ON s.ProductKey = p.ProductKey
GROUP BY p.Brand
ORDER BY Total_Quantity_Sold DESC;
"""

pd.read_sql(query, connection)
```

### Business Insight

This analysis identifies the brands with the highest sales volume. High-performing brands contribute significantly to business growth and should be prioritized for inventory planning and marketing campaigns.

## Business Question 3

### Which product categories generate the highest sales volume?

```python
query = """
SELECT
    p.Category,
    SUM(s.Quantity) AS Total_Quantity_Sold
FROM sales s
INNER JOIN products p
ON s.ProductKey = p.ProductKey
GROUP BY p.Category
ORDER BY Total_Quantity_Sold DESC;
"""

pd.read_sql(query, connection)
```

### Business Insight

Product categories with higher sales volumes represent the company's strongest revenue drivers. These insights support inventory optimization and strategic product planning.

## Business Question 4

### Which product categories receive the highest number of orders?

```python
query = """
SELECT
    p.Category,
    COUNT(DISTINCT s."Order Number") AS Total_Orders
FROM sales s
INNER JOIN products p
ON s.ProductKey = p.ProductKey
GROUP BY p.Category
ORDER BY Total_Orders DESC;
"""

pd.read_sql(query, connection)
```

### Business Insight

This analysis measures customer demand by counting distinct orders for each product category, providing insights beyond sales quantity alone.

## Business Question 5

### Which brands sold the highest number of products?

```python
query = """
SELECT
    p.Brand,
    SUM(s.Quantity) AS Products_Sold
FROM sales s
INNER JOIN products p
ON s.ProductKey = p.ProductKey
GROUP BY p.Brand
ORDER BY Products_Sold DESC
LIMIT 10;
"""

pd.read_sql(query, connection)
```

### Business Insight

The leading brands contribute the largest share of products sold. These brands are essential for maintaining customer demand and business performance.
