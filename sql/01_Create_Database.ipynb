# Create SQLite Database

In this notebook, we create a SQLite database using the cleaned datasets.

The objective is to simulate a relational database environment commonly used in business analytics.

```python
import pandas as pd
import sqlite3
```

```python
connection = sqlite3.connect("../sql/global_electronics.db")

print("Database Created Successfully")
```

```python
customers = pd.read_csv("../data/Customers.csv", encoding="cp1252")
products = pd.read_csv("../data/Products.csv", encoding="cp1252")
sales = pd.read_csv("../data/Sales.csv", encoding="cp1252")
stores = pd.read_csv("../data/Stores.csv", encoding="cp1252")
exchange = pd.read_csv("../data/Exchange_Rates.csv", encoding="cp1252")
```

```python
customers.to_sql(
    "customers",
    connection,
    if_exists="replace",
    index=False
)

products.to_sql(
    "products",
    connection,
    if_exists="replace",
    index=False
)

sales.to_sql(
    "sales",
    connection,
    if_exists="replace",
    index=False
)

stores.to_sql(
    "stores",
    connection,
    if_exists="replace",
    index=False
)

exchange.to_sql(
    "exchange_rates",
    connection,
    if_exists="replace",
    index=False
)

print("All Tables Created Successfully")
```

```python
query = """
SELECT name
FROM sqlite_master
WHERE type='table';
"""

pd.read_sql(query, connection)
```

```python
tables = [
    "customers",
    "products",
    "sales",
    "stores",
    "exchange_rates"
]

for table in tables:

    count = pd.read_sql(
        f"SELECT COUNT(*) AS Rows FROM {table}",
        connection
    )

    print(table)
    print(count)
    print("-"*30)
```

```python
connection.close()

print("Database Saved Successfully")
```
