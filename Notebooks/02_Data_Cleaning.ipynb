# Data Cleaning & Preprocessing

## Objective

The objective of this notebook is to clean and preprocess the Global Electronics Retail dataset before performing SQL analysis and building the Power BI dashboard.

### Tasks
- Load all datasets
- Inspect structure and data types
- Handle missing values
- Remove duplicates
- Standardize column names
- Convert data types
- Save cleaned datasets

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 10)
```

## Load the Datasets

```python
customers = pd.read_csv("../data/Customers.csv", encoding="cp1252")
products = pd.read_csv("../data/Products.csv", encoding="cp1252")
sales = pd.read_csv("../data/Sales.csv", encoding="cp1252")
stores = pd.read_csv("../data/Stores.csv", encoding="cp1252")
exchange = pd.read_csv("../data/Exchange_Rates.csv", encoding="cp1252")
```

## Dataset Dimensions

```python
datasets = {
    "Customers": customers,
    "Products": products,
    "Sales": sales,
    "Stores": stores,
    "Exchange Rates": exchange
}

for name, df in datasets.items():
    print("=" * 40)
    print(name)
    print(df.shape)
```

## Check Data Types

```python
for name, df in datasets.items():
    print("\n", "=" * 50)
    print(name)
    print(df.dtypes)
```

## Check Missing Values

```python
for name, df in datasets.items():
    print("\n", "=" * 50)
    print(name)
    print(df.isnull().sum())
```

## Remove Duplicate Records

Duplicate records can lead to inaccurate analysis and incorrect business insights. In this step, we identify and remove duplicate rows from each dataset.

```python
for name, df in datasets.items():
    duplicates = df.duplicated().sum()
    print(f"{name}: {duplicates} duplicate rows")
```

## Standardize Column Names

To maintain consistency across datasets, column names are converted to lowercase and spaces are replaced with underscores.

```python
for df in datasets.values():
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
    )
```

## Dataset Information After Cleaning

```python
for name, df in datasets.items():
    print("=" * 60)
    print(name)
    print(df.info())
```

## Summary Statistics

```python
for name, df in datasets.items():
    print("=" * 60)
    print(name)
    display(df.describe(include="all"))
```

## Save Cleaned Datasets

After cleaning, the datasets are saved for SQL analysis and dashboard development.

```python
customers.to_csv("../data/customers_cleaned.csv", index=False)
products.to_csv("../data/products_cleaned.csv", index=False)
sales.to_csv("../data/sales_cleaned.csv", index=False)
stores.to_csv("../data/stores_cleaned.csv", index=False)
exchange.to_csv("../data/exchange_rates_cleaned.csv", index=False)
```
