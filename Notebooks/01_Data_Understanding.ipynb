# Business Data Model

The Global Electronics Retailer dataset follows a relational database model.

## Tables

### Customers
Stores customer demographic and location information.

**Primary Key**
- CustomerKey

---

### Products
Contains information about electronic products.

**Primary Key**
- ProductKey

---

### Stores
Contains store information such as country, state, square meters, and opening date.

**Primary Key**
- StoreKey

---

### Sales
Stores every sales transaction.

**Primary Key**
- Order Number (or OrderNumber if available)

**Foreign Keys**
- CustomerKey
- ProductKey
- StoreKey
- Currency Code

---

### Exchange Rates
Contains daily exchange rates for different currencies.

**Primary Key**
- Date + Currency

Used to convert sales into a common reporting currency.

        ▼
 Exchange Rates

# Business Workflow

The business process can be summarized as follows:

1. A customer visits a store.
2. The customer purchases one or more products.
3. Each purchase is recorded in the Sales table.
4. Every sale references:
   - Customer
   - Product
   - Store
   - Currency
5. Exchange rates are used to standardize financial reporting across countries.
